```python
import os
import pandas as pd

# Define the path to your raw data folder
RAW_DATA_DIR = r"../data/"

# The specific files we need for our Anchor Cohort strategy
expected_files = [
    "2000Q1.csv", 
    "2005Q1.csv", 
    "2008Q1.csv", 
    "2013Q1.csv", 
    "2020Q1.csv"
]

print("🔍 PHASE 1: Raw Data Verification Protocol\n" + "="*50)

for file_name in expected_files:
    file_path = os.path.join(RAW_DATA_DIR, file_name)
    
    # 1. Check if the file exists
    if not os.path.exists(file_path):
        print(f"❌ MISSING: {file_name} not found in {RAW_DATA_DIR}")
        continue
        
    # 2. Check the file size (These should be massive, usually 500MB - 2GB)
    file_size_mb = os.path.getsize(file_path) / (1024 * 1024)
    
    # 3. Read JUST the first row to verify the delimiter and column count
    try:
        # We read 1 row, forcing the pipe separator
        sample_df = pd.read_csv(file_path, sep='|', header=None, nrows=1)
        col_count = sample_df.shape[1]
        
        # Fannie Mae performance files typically have 100+ columns
        if col_count > 100:
            print(f"✅ VERIFIED: {file_name} | Size: {file_size_mb:.1f} MB | Columns: {col_count}")
        else:
            print(f"⚠️ WARNING: {file_name} | Only found {col_count} columns. Delimiter issue?")
            
    except Exception as e:
        print(f"❌ ERROR reading {file_name}: {e}")

print("="*50)
print("If all files are marked ✅ VERIFIED, Phase 1 is complete.")
```
